# VISUALIZATION — CHURN ROOT-CAUSE ANALYSIS

```python
import pandas as pd
import matplotlib.pyplot as plt 
```

## CHURNED VS ACTIVE SIGNAL COMPARISON

```python


OUTPUT_DIR = r"D:\mySQL\Project\outputs"

```

```python




comparison = pd.read_csv(
    f"{OUTPUT_DIR}/churn_signal_comparison.csv"
)

plot_df = comparison.set_index("signal")[
    ["churned_pct", "active_pct"]
]

ax = plot_df.plot(
    kind="bar",
    figsize=(10, 6)
)

ax.set_title(
    "Churned vs Active Users — Key Churn Signals"
)

ax.set_xlabel("Signal")
ax.set_ylabel("Users (%)")

plt.xticks(rotation=0)
plt.legend(
    ["Churned Users", "Active Users"],
    title="Customer Status"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/churn_signal_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


```

## USAGE DROP MAGNITUDE

```python


drop_df = pd.read_csv(
    f"{OUTPUT_DIR}/usage_drop_magnitude_comparison.csv"
)

# Pivot for side-by-side comparison
drop_plot = (
    drop_df
    .pivot(
        index="drop_bucket",
        columns="status",
        values="user_count"
    )
    .fillna(0)
)

# Keep buckets in logical order
bucket_order = [
    "0-10% drop",
    "10-20% drop",
    "20-30% drop",
    "30-40% drop",
    "40-50% drop",
    "Severe (>50%) drop"
]

drop_plot = drop_plot.reindex(bucket_order)

ax = drop_plot.plot(
    kind="bar",
    figsize=(11, 6),
    
)

ax.set_title(
    "Usage Decline Magnitude — Churned vs Active Users"
)

ax.set_xlabel("Usage Decline")
ax.set_ylabel("Number of Users")

plt.xticks(rotation=20)
plt.legend(
    title="Customer Status"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/usage_drop_magnitude_comparison.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()
```
